## Data Cleaning Summary

The dataset was inspected using Pandas to understand its dimensions, column names, data types, and missing values.

The initial dataset contained 11,251 records and 15 columns. The Status and unnamed1 columns had 100% missing values and were removed. The Amount column contained 12 missing values, which were imputed using the median value of 8,109. The Amount column was then converted to an integer data type.

After removing the two unnecessary columns and handling missing Amount values, the dataset contained 13 columns and no remaining missing values.

This preparation made the dataset more suitable for exploratory analysis.

# Exploratory Data Analysis

### Gender analysis

```python
df.columns
```

```python
ax = sns.countplot(x='Gender',data=df,palette='Set2',hue='Gender',legend=False)
for bars in ax.containers:
    ax.bar_label(bars)
plt.show()
```

```python
df.groupby(['Gender'],as_index=False)['Amount'].sum().sort_values(by='Amount',ascending=False)
```

```python
sales_gender = (df.groupby('Gender', as_index=False)['Amount'].sum()
.sort_values(by='Amount', ascending=False))
ax = sns.barplot(x='Gender',y='Amount',data=sales_gender,hue='Gender',palette='Set2',legend=False)
plt.title('Total Sales by Gender')
plt.xlabel('Gender')
plt.ylabel('Total Amount')
plt.show()
```

### from above graphs we can see that most of the buyers are females and even the purchasing power of females are greater than man 

# Age

```python
df.columns
```

```python
ax = sns.countplot(data=df,x='Age Group',hue='Gender',palette={'M': 'cornflowerblue', 'F': 'lightcoral'})
for bars in ax.containers:
    ax.bar_label(bars)
plt.title('Age Group Distribution by Gender')
plt.xlabel('Age Group')
plt.ylabel('Count')
plt.legend(title='Gender', labels=['Female', 'Male'])

plt.show()
```

```python
sales_age = (df.groupby('Age Group', as_index=False)['Amount'].sum().sort_values(by='Amount', ascending=False))
plt.figure(figsize=(10, 5))
ax = sns.barplot(data=sales_age,x='Age Group',y='Amount',hue='Age Group',palette='Set2',legend=False)
for bars in ax.containers:
    ax.bar_label(bars, fmt='%.0f')
plt.title('Total Sales by Age Group')
plt.xlabel('Age Group')
plt.ylabel('Total Sales Amount')
plt.tight_layout()

plt.show()
```

# Marital Status

```python
ax = sns.countplot(data=df, x='Marital_Status', hue='Marital_Status', palette='Set2', legend=False)
for bars in ax.containers:
    ax.bar_label(bars)
plt.show()
```

```python
ax = sns.countplot(data=df,x='Marital_Status',hue='Gender',palette={'F': 'lightcoral', 'M': 'cornflowerblue'})
for bars in ax.containers:
    ax.bar_label(bars)
plt.title('Marital Status Distribution by Gender')
plt.xlabel('Marital Status (0 = Unmarried, 1 = Married)')
plt.ylabel('Count')
plt.legend(title='Gender', labels=['Female', 'Male'])
plt.show()
```

# Occupation

```python
plt.figure(figsize=(12, 5))
ax = sns.countplot(data=df, x='Occupation', hue='Occupation', palette='Set2', legend=False)
for bars in ax.containers:
    ax.bar_label(bars)

plt.title('Customer Count by Occupation')
plt.xticks(rotation=45)
plt.show()
```

###  from above we can see that most of the buyers are working in IT,Healthcare and Aviation Sector

# Product Category

```python
plt.figure(figsize=(12, 5))
ax = sns.countplot(data=df,x='Product_Category',hue='Product_Category',palette='Set2',legend=False)
for bars in ax.containers:
    ax.bar_label(bars)

plt.title('Product Category Count')
plt.xlabel('Product Category')
plt.ylabel('Count')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
sales_product = df.groupby('Product_Category')['Amount'].sum()
ax = sales_product.plot(kind='bar', figsize=(12, 5), color='teal')

for bars in ax.containers:
    ax.bar_label(bars, fmt='%.0f', fontsize=8)

plt.title('Total Spending by Product Category', fontsize=14)
plt.xlabel('Product Category')
plt.ylabel('Total Amount Spent')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

###  from above graphs we can see hat most of the sold products are Food,Clothing and Electronics category 

```python
top10 = df['Product_ID'].value_counts().head(10)
top10.plot(kind='bar', color='teal', figsize=(10, 5))
plt.title('Top 10 Products Sold')
plt.xlabel('Product ID')
plt.ylabel('Number of Sales')
plt.xticks(rotation=45)
plt.show()
```
